## 1. Extended Thinking

Better reasoning capabilities for more complez tasks (aka reasoning).

-> Thinking consumes more tokens: increased cost and latency.

-> Use evals to decide if you need extended thinking: write a prompt (a great prompt), run evals, use extended thinking if accuracy is not good enough.


|  | Without thinking | With thinking |
| -- | -- | -- |
| User Message: Text Block | yes | yes |
| Assistant Message: Thinking Block | no | yes |
| Assistant Message: Text Block | yes | yes |

**Signature**: Cryptographic token tied to the thibnking text that Claude generated. Verifies that the exact text was generated by Claude, and not changed

**Redacted thinking**: Reasoning text might be flagged by Claude's internal safety systems. Provided to hand back to Claude on follow up requests

*Important Note: Extended Thinking is not compatible with some other features, notable message pre-filling and temperature. See the full list of restrictions [here](https://docs.anthropic.com/en/docs/build-with-claude/extended-thinking#feature-compatibility).* 

````python
def chat(
    messages,
    system=None,
    temperature=1.0, 
    stop_sequences=[],
    max_tokens=4000,
    tools=None,
    thinking=False,
    thinking_budget=1024,
    client: Anthropic=client
):
    params = {
        "model": model,
        "max_tokens": max_tokens,
        "messages": messages,
        "temperature": temperature,
        "stop_sequences": stop_sequences,
    }

    if thinking:
        params["temperature"] = 1.0 # Temperature needs to be set to 1.0 for thinking to work
        params["thinking"] = {
            "type": "enabled",
            "budget_tokens": thinking_budget,
        }

    if tools:
        params["tools"] = tools

    if system:
        params["system"] = system

    message = client.messages.create(**params)
    return message


def text_from_message(message):
    return "\n".join([block.text for block in message.content if block.type == "text"])
        
````

## 4. Prompt Caching

When receiving an input, Claude: tokenizes the prompt, creates embeddings for each token, adds context based on surrounding text, generates output text... and it's then deleted from memory. Any follow-up requests to the original input will require Claude to do it all again.

-> Instead, there can be some temporary chaching done for the original input message.


* **Faster responses**: Requests using cached content execute more quickly
* **Lower costs**: You pay less for the cached portions of your requests
* **Automatic optimization**: The initial request writes to the cache, follow-up requests read from it 
* Cached content only lives for **one hour**.
* Only useful if you're repeatedly sending the same content, but this is *extremely* frequent:
    * document analysis
    * multiple questions about the same large document
    * iterative editing tasks (same base content, specific aspects refined.)

**Minimum content length**: Content must be at least 1024 tokens long to be cached.

##### Cache Breakpoints

Chaching is not automatic. It needs to be included in a text block sent to Claude.
When including a breakpoint, all the content up to, and including that breakpoint, will be cached. Content after the breakpoint is not cached.

**Important**: For the cache to be useful, the content must be **identical** up to the breakpoint (a.k.a. all messages are the same). Any tiny difference will invalidate the cache and all content will be reprocessed.

Ex.:

```
user_message = {
    "role": "user",
    "content":[
        {
            "type": "test",
            "text": "<Long Prompt>",
            "cache_control"{
                "type": "ephemeral"
            }
        }
    ]
}
```

You're not limited to text blocks - by adding a `cache_control` item, cache breakpoints can be added to:

* System prompts
* Tool definitions
* Image blocks
* Tool use and tool result blocks

*Tool definitions and system prompts are great candidates for caching (they rarely change between requests).*

##### Cache Ordering

Blocks are always cached in the same order:
1. Tool #1
2. Tool #2
3. Tool #3 (**Cache Breakpoint**)
4. System Prompt
5. User Message
6. Assistant Message (**Cache Breakpoint**)
7. User Message

-> If User/Assistant messages change, all tools will be cached anyways.
-> There can be multiple cache breakpoints

#### Code

````python
# Helper functions
from anthropic.types import Message

def add_user_message(messages, message):
    user_message = {
        "role": "user",
        "content": message.content if isinstance(message, Message) else message,
    }
    messages.append(user_message)

def add_assistant_message(messages, message):
    assistant_message = {
        "role": "assistant",
        "content": message.content if isinstance(message, Message) else message,
    }
    messages.append(assistant_message)

def chat(
    messages,
    system=None,
    temperature=1.0,
    stop_sequences=[],
    tools=None,
    thinking=False,
    thinking_budget=1024,
    ):
    params = {
        "model": model,
        "max_tokens": 4000,
        "messages": messages,
        "temperature": temperature,
        "stop_sequences": stop_sequences,
    }

    if thinking:
        params["thinking"] = {
            "type": "enabled",
            "budget_tokens": thinking_budget,
        }

    if tools:
        # always cache the last tool
        # not the best but would do: 
        # tools[-1]["cache_control"] = {"type": "ephemeral"} -> changes this specific tool definition
        # Better, in case the list of tools is /reordered:
        tools_clone = tools.copy()
        last_tool = tools_clone[-1].copy()
        last_tool["cache_control"] = {"type": "ephemeral"}
        tools_clone[-1] = last_tool
        params["tools"] = tools_clone

    if system:
        # always cache the system prompt
        params["system"] = [
            {
                "type": "text",
                "text": system,
                "cache_control": {"type": "ephemeral"}
            }
        ]

    message = client.messages.create(**params)
    return message

def text_from_message(message):
    return "\n".join([block.text for block in message.content if block.type == "text"])
````

## 5. Code execution and the Files API

**Files API:** Alternative way to handle file uploads: instead of encoding files in messages as base64 data, they can be uploaded and reference them later (particularly useful for heavy files).

1. Upload your file (image, PDF, text, etc.) to Claude using a separate API call
2. Receive a file metadata object containing a unique file ID
3. Reference that file ID in future messages instead of including raw file data


**Code execution:** Claude can execute code in a Docker container: provide a predefined tool schema and Claude can execute it in an isolated environment.


* Upload via `container upload block`.
- No need to provide implementation.
- No network access (no external API calls).
- Claude can execute code multiple times during a single conversation.
- Results are captured and interpreted for final response.


**Combining features**: since there's no network access, Claude can get data in/out of the execution environment via `Files API`.

Workflow:
1. Upload your data file (like a CSV) using the Files API.
2. Include a container upload block in your message with the file ID.
3. Ask Claude to analyze the data.
4. Claude writes and executes code to process your file.
5. Claude can generate outputs (like plots) that you can download.


````python
# Declare new client with code execution and files API enabled
client2 = Anthropic(
    default_headers={
        "anthropic-beta": "code-execution-2025-08-25, files-api-2025-04-14"
    }
)   
````

````python
# Additional helper functions
from pathlib import Path
# Note: as of 14/10/2026, the temperature parameter has been deprecated for all models released after CLaude Opus 4.6.

def chat(
    messages,
    system=None,
    #temperature=1.0, 
    stop_sequences=[],
    max_tokens=4000,
    tools=None,
    thinking=False,
    thinking_budget=1024,
    client: Anthropic=client2
    ):
    params = {
        "model": model,
        "max_tokens": max_tokens,
        "messages": messages,
        #"temperature": temperature,
        "stop_sequences": stop_sequences,
    }

    if thinking:
        #params["temperature"] = 1.0 # Temperature needs to be set to 1.0 for thinking to work
        params["thinking"] = {
            "type": "enabled",
            "budget_tokens": thinking_budget,
        }

    if tools:
        params["tools"] = tools

    if system:
        params["system"] = system

    message = client.messages.create(**params)
    return message

def upload(file_path, client: Anthropic = client2):
    path = Path(file_path)
    extension = path.suffix.lower()

    mime_type_map = {
        ".pdf": "application/pdf",
        ".txt": "text/plain",
        ".md": "text/plain",
        ".py": "text/plain",
        ".js": "text/plain",
        ".html": "text/plain",
        ".css": "text/plain",
        ".csv": "text/csv",
        ".json": "application/json",
        ".xml": "application/xml",
        ".xlsx": "application/vnd.openxmlformats-officedocument.spreadsheetml.sheet",
        ".xls": "application/vnd.ms-excel",
        ".jpeg": "image/jpeg",
        ".jpg": "image/jpeg",
        ".png": "image/png",
        ".gif": "image/gif",
        ".webp": "image/webp",
    }

    mime_type = mime_type_map.get(extension)

    if not mime_type:
        raise ValueError(f"Unknown mimetype for extension: {extension}")
    filename = path.name

    with open(file_path, "rb") as file:
        # Returns File ID
        return client.beta.files.upload(file=(filename, file, mime_type))

def list_files(client:Anthropic = client2):
    return client.beta.files.list()

def delete_file(id, client:Anthropic = client2):
    return client.beta.files.delete(id)

def download_file(id, filename=None, client:Anthropic = client2):
    file_content = client.beta.files.download(id)

    if not filename:
        file_metadata = get_metadata(id)
        file_content.write_to_file(file_metadata.filename)
    else:
        file_content.write_to_file(filename)

def get_metadata(id, client:Anthropic = client2):
    return client.beta.files.get(id)
````
